```python
PROJECT_PATH = "/content/drive/MyDrive/PTB-XL Research Project"

DATA_PATH = os.path.join(
    PROJECT_PATH,
    "data",
    "ptb-xl-a-large-publicly-available-electrocardiography-dataset-1.0.3"
)

print("Project path:")
print(PROJECT_PATH)

print("\nDataset path:")
print(DATA_PATH)

print("\nDataset folder exists:")
print(os.path.isdir(DATA_PATH))
```

```python
database_path = os.path.join(
    DATA_PATH,
    "ptbxl_database.csv"
)

scp_path = os.path.join(
    DATA_PATH,
    "scp_statements.csv"
)

df = pd.read_csv(
    database_path,
    index_col="ecg_id"
)

scp = pd.read_csv(
    scp_path,
    index_col=0
)

print("ECG records:", len(df))
print("ECG columns:", len(df.columns))
print("SCP statements:", len(scp))
```
